```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 22],
    "Marks": [85, 78, 90]
}

df = pd.DataFrame(data)

# Access row using loc
print("Using loc:")
print(df.loc[1])

# Access row using iloc
print("\nUsing iloc:")
print(df.iloc[1])
```

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Marks": [85, 78, 90, 65]
}

df = pd.DataFrame(data)

# Filter students with marks > 80
result = df[df["Marks"] > 80]

print(result)
```

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", None],
    "Age": [20, None, 22],
    "Marks": [85, 78, None]
}

df = pd.DataFrame(data)

# Display DataFrame
print("DataFrame:")
print(df)

# Identify missing values
print("\nMissing Values:")
print(df.isnull())
```

```python
import pandas as pd

# Create a DataFrame with missing values
data = {
    "Name": ["Alice", "Bob", None],
    "Age": [20, None, 22],
    "Marks": [85, 78, None]
}

df = pd.DataFrame(data)

# Replace missing values with 0
df = df.fillna(0)

# Display the updated DataFrame
print(df)
```
